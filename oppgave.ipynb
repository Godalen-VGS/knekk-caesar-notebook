I oppgave 13 krypterte og dekrypterte du med Cæsarchifferet når du **visste** forskyvningen. Nå skal du knekke chifferet **uten** å vite den, slik en kodeknekker ville gjort.

Funksjonen `krypter` under er ferdig laget for deg (den flytter bare a–z, så æ, ø og å står urørt). Kjør cellen før du begynner.

> Behold funksjonsnavnet `knekk(kryptert_tekst)` nøyaktig slik. Det er det som gjør at selvtesten og den automatiske rettingen kan sjekke svarene dine.

```python
def krypter(tekst, skift):
    """Krypterer tekst med Cæsarchifferet. Bare a–z og A–Z flyttes, alt annet beholdes."""
    resultat = ""
    for tegn in tekst:
        if "a" <= tegn <= "z":
            resultat += chr((ord(tegn) - ord("a") + skift) % 26 + ord("a"))
        elif "A" <= tegn <= "Z":
            resultat += chr((ord(tegn) - ord("A") + skift) % 26 + ord("A"))
        else:
            resultat += tegn
    return resultat
```
